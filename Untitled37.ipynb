```python
# Creating a separate modelling dataset

model_df = df.copy()

# Removing duration to prevent data leakage
model_df = model_df.drop(columns=['duration'])

# Separating features and target

X = model_df.drop(columns=['y'])
y = model_df['y']

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())
```

```python
# Identifying feature types

numerical_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X.select_dtypes(
    include='object'
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
# Machine Learning imports

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)

print("Machine Learning libraries imported successfully!")
```

```python
# Splitting the dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
# Creating preprocessing pipeline

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),

        ('cat', OneHotEncoder(handle_unknown='ignore'),
         categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
# Creating Logistic Regression pipeline

logistic_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),

        ('classifier', LogisticRegression(
            class_weight='balanced',
            max_iter=2000,
            random_state=42
        ))
    ]
)

# Training the model

logistic_model.fit(X_train, y_train)

print("Logistic Regression Model Trained Successfully!")
```

```python
# Predictions

y_pred = logistic_model.predict(X_test)

# Prediction probabilities

y_prob = logistic_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")

print("\nFirst 10 Actual Values:")
print(y_test.iloc[:10].values)

print("\nFirst 10 Predicted Values:")
print(y_pred[:10])
```

```python
# Model evaluation

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(y_test, y_pred)

recall = recall_score(y_test, y_pred)

f1 = f1_score(y_test, y_pred)

roc_auc = roc_auc_score(y_test, y_prob)

print("LOGISTIC REGRESSION PERFORMANCE")
print("--------------------------------")

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
print("ROC-AUC  :", round(roc_auc, 4))
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=['Not Subscribed', 'Subscribed']
    )
)
```

```python
# Confusion Matrix

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['No', 'Yes']
)

disp.plot(cmap='Blues')

plt.title('Logistic Regression - Confusion Matrix')
plt.show()
```

```python
# ROC Curve

RocCurveDisplay.from_predictions(
    y_test,
    y_prob
)

plt.title('Logistic Regression - ROC Curve')
plt.show()
```
