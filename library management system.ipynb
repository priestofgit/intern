```python
books =[]
```

```python

def add_book():
    book_id = input("enter book id: ")
    title = input("enter book title: ")
    author_name = input("enter author name: ")
    number = int(input("enter number of books: "))
    book = {
        "book_id": book_id,
        "title": title,
        "author_name": author_name,
        "number": number
    }
    books.append(book)
    print("book added successfully")

def book_display():
    if not books:
        print("No books available")
        return
    for book in books:
        print("book_id:", book["book_id"])
        print("title:", book["title"])
        print("author_name:", book["author_name"])
        print("number:", book["number"])
        
def search_book():
    book_id = input("enter book_id to search: ")
    for book in books:
        if book["book_id"] == book_id:
            print("book Found!")
            print("title:", book["title"])
            print("author_name:", book["author_name"])
            print("number:", book["number"])
            return

    print("book not found!")

def issue_book():
    book_id = input("Enter book_id to issue: ")

    for book in books:
        if book["book_id"] == book_id:
            if book["number"] > 0:
                student = input("Enter Student Name: ")
                book["number"] -= 1
                print("Book issued successfully to", student)
            else:
                print("Book is not available!")
            return

    print("Book not found!")
    
def main():
    while True:
        print("\n========== LIBRARY MANAGEMENT SYSTEM ==========")
        print("1. Add Book")
        print("2. Display Books")
        print("3. Search Book")
        print("4. Issue Book")
        print("5. Exit")
        choice = input("\nEnter your choice: ")
        if choice == "1":
            add_book()
        elif choice == "2":
            book_display()
        elif choice == "3":
            search_book()
        elif choice == "4":
            issue_book()
        elif choice == "5":
            print("\nThank you for using Library Management System!")
            break
        else:
            print("\nInvalid choice! Try again.")
main()
```
